```python
import json
import os
```

```python
output_directory = '/Users/aadi.joshi/Desktop/scientific_fact_checking/tabular_scientific_fact_checking/code/detection/output/detect_gpt54_workspace'
batch_files = []

for filename in os.listdir(output_directory):
    if filename.startswith("batch_batch_") and filename.endswith(".json"):
        batch_files.append(os.path.join(output_directory, filename))
```

```python
results = []

for batch_file in batch_files:
    with open(batch_file, 'r') as f:
        for line in f:
            data = json.loads(line)
            results.append(data)
```

```python
len(set([r['custom_id'].split('___')[0] for r in results]))
```

```python
results[0]
```

```python
results_by_input = {
    "images": [],
    "latex": [],
    "images_and_latex": []
}

for res in results:
    input_type = res['custom_id'].split("___")[-1]
    results_by_input[input_type].append(res)
```

```python
results_by_input['images'][0]
```

```python
results_by_input['images'][0]['response']['body']['usage']
```

```python
for input_type, res_list in results_by_input.items():
    usage = [res['response']['body']['usage'] for res in res_list]
    total_input_tokens = sum(u['input_tokens'] for u in usage)
    total_cached_tokens = sum(u['input_tokens_details']['cached_tokens'] for u in usage)
    total_output_tokens = sum(u['output_tokens'] for u in usage)
    total_reasoning_tokens = sum(u['output_tokens_details']['reasoning_tokens'] for u in usage)
    avg_input_tokens = total_input_tokens / len(res_list) if res_list else 0
    avg_output_tokens = total_output_tokens / len(res_list) if res_list else 0
    print(f"Input Type: {input_type}")
    print(f"  Total Samples: {len(res_list)}")
    print(f"  Total Input Tokens: {total_input_tokens}")
    print(f"      Cached Tokens: {total_cached_tokens}")
    print(f"  Total Output Tokens: {total_output_tokens}")
    print(f"      Reasoning Tokens: {total_reasoning_tokens}")
    print(f"  Max Input Tokens: {max(u['input_tokens'] for u in usage) if usage else 0}")
    print(f"  Max Output Tokens: {max(u['output_tokens'] for u in usage) if usage else 0}")
    print(f"  Average Input Tokens: {avg_input_tokens:.2f}")
    print(f"  Average Output Tokens: {avg_output_tokens:.2f}")
```
