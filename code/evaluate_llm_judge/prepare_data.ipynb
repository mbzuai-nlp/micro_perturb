```python
import dill
import pandas as pd
```

```python
MODEL_ID = 'gpt_54'
INPUT_TYPE = 'latex'
```

```python
with open('responses_judge_gpt_54.dill', 'rb') as fp:
    responses = dill.load(fp)
```

```python
data = []
for paper_id, err_id, res in responses:
    # not validated so never created the ground truth
    if 'ground_truth' not in res:
        continue
    
    ground_truth = res['ground_truth'][0]['response_text']

    model_generated_res = ''
    for model_res in res['model_responses']:
        if model_res['model_id'] == MODEL_ID:
            model_generated_res = model_res['response_text']
            break

    data.append((paper_id, err_id, MODEL_ID, INPUT_TYPE, ground_truth, model_generated_res))
```

```python
df = pd.DataFrame(data, columns=['paper_id', 'error_id', 'model_id', 'input_type', 'ground_truth', 'model_response'])
```

```python
df.to_csv(f'judge_evaluation_{MODEL_ID}_{INPUT_TYPE}.csv')
```
